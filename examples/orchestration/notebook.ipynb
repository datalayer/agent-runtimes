# Quarterly revenue, by region

The notebook the orchestration example analyses. It has real problems in
it, so that a worker reporting "looks fine" would be visibly wrong.

```python
import pandas as pd
```

```python
revenue = pd.DataFrame(
    {
        "region": ["emea", "amer", "apac", "emea", "amer", "apac"],
        "quarter": ["Q1", "Q1", "Q1", "Q2", "Q2", "Q2"],
        "revenue": [412_000, 690_500, 231_900, 448_100, 712_300, 205_400],
    }
)
revenue.head()
```

```python
# The column is `revenue`, not `total_revenue`.
by_region = revenue.groupby("region")["total_revenue"].sum()
by_region
```

```python
growth = by_region.pct_change()
print(f"quarter-on-quarter growth: {growth}")
```

```python
import matplotlib.pyplot as plt

by_region.plot(kind="bar")
plt.show()
```

Conclusion: revenue grew in every region.
